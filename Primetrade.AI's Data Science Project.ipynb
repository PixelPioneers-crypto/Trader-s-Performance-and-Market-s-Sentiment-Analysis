```python
# Merging the Datasets to answer Trader's Behavior with the Market Sentiment

Merged_Dataset = df1.merge(df[['date', 'classification', 'value']], on = 'date', how = 'left') # Answer to why only 'Left' Join: The trading dataset 

# is the primary dataset. We want to analyze: How does a trader's behavior change depending on market sentiment?

# Therefore: Every trade must be preserved.
```

```python
Merged_Dataset
```

```python
# Daily PnL per trader (or per account)
Daily_pnl = Merged_Dataset.groupby(['Account','date'])['Closed PnL'].sum().reset_index()
```

```python
Daily_pnl
```

```python
# Win rate

Merged_Dataset['Win'] = Merged_Dataset['Closed PnL'] > 0

Win_Rate = Merged_Dataset.groupby('Account')['Win'].mean().reset_index()
Win_Rate.rename(columns={'Win':'Win_Rate'}, inplace = True) # inplace = True is used to make the changes permanent.
```

```python
Win_Rate
```

```python
# Average Trade Size

Avg_Trade_Size = Merged_Dataset.groupby('Account')['Size USD'].mean().reset_index()
```

```python
Avg_Trade_Size
```

```python
# Number of trades per day

Trades_per_day = Merged_Dataset.groupby('date')['Trade ID'].count().reset_index()
Trades_per_day.rename(columns = {'Trade ID':'Trades_in_a_single_day'}, inplace = True)
```

```python
Trades_per_day
```

```python
# Long / Short Ratio

# fill_value = 0: Ensures missing BUY/SELL values become 0 instead of NaN.

# replace(0, 1): Prevents division by zero errors.

long_short = ( Merged_Dataset.groupby(['date','Side']).size().unstack(fill_value = 0))

long_short['long_short_ratio'] = long_short['BUY'] / long_short['SELL'].replace(0, 1)
```

```python
# Q.) Does performance differ between Fear vs Greed days?

Performance_difference = Merged_Dataset.groupby('classification')['Closed PnL'].mean()
```

```python
sns.boxplot(x = 'classification', y = 'Closed PnL', data = Merged_Dataset)
plt.title("Profit n Loss Distribution by Market Sentiment")
plt.show()
```

```python
# Q.) Do traders change behavior based on sentiment?

# agg is the function used when we need to perform the Aggregate functions on the multiple columns.

Behaviour = Merged_Dataset.groupby('classification').agg({
'Size USD':'mean',
'Trade ID':'count',
'Closed PnL':'mean'
})
```

```python
Behaviour
```

```python
# Q.) Trader Segmentation

# Segment 1 — High vs Low Leverage Traders

Merged_Dataset['Leverage_segment'] = pd.qcut(Merged_Dataset['Size USD'], 2 ,labels = ['Low','High']) # It takes your 'Size US' data, 
# sorts it, splits it in half, and labels the smaller companies "Low" and the larger companies "High". 


# Segment 2 — Frequent vs Infrequent Traders

Trader_activity = Merged_Dataset.groupby('Account').size()

Threshold = Trader_activity.median()

Segments = Trader_activity.apply(
lambda x: "Frequent" if x > Threshold else "Infrequent"
)

# Segment 3 — Consistent vs Inconsistent Traders

pnl_std = Merged_Dataset.groupby('Account')['Closed PnL'].std()
```

```python
# Visual Representation of any of the 3 Charts.

plt.figure(figsize = (10,6))

sns.boxplot(
    data = Merged_Dataset,
    x = 'classification',
    y = 'Closed PnL'
)

plt.title("Trader Performance Across the Market")
plt.xlabel("Market Sentiment")
plt.ylabel("Closed PnL")
plt.show()
```

```python
Activity = Merged_Dataset.groupby('classification').size().reset_index(name = 'Trade_count')

sns.barplot(
    data = Activity,
    x = 'classification',
    y = 'Trade_count',
    errorbar = None
)
```

```python
# Vertical lines on each barplot of the visualization are the whiskers present, i.e., they show the spread of trade sizes.

plt.figure(figsize = (10,6))

sns.barplot(
    data = Merged_Dataset,
    x = 'classification',
    y = 'Size USD',
   )

plt.title("Trade Size Distribution Across Market Sentiment", fontsize = 14)
plt.xlabel("Market Sentiment", fontsize = 12)
plt.ylabel("Position Size (USD)", fontsize = 12)

plt.tight_layout() # For the better spacing
plt.show()
```
